```python
txx = xr.open_dataset(
    BASE / "output" / "txx" / "TXx_annual_1950_2100.nc"
)

txx
```

```python
txx = txx.rio.write_crs("EPSG:4326")

txx
```

```python
txx = txx.rio.set_spatial_dims(
    x_dim="lon",
    y_dim="lat"
)
```

```python
txx_clip = txx.rio.clip(
    shape.geometry,
    shape.crs,
    drop=True
)
```

```python
outfile = BASE / "output" / "txx" / "TXx_VUT_clipped.nc"

txx_clip.to_netcdf(outfile)
```

```python
def clip_netcdf(ds, shp):

    ds = ds.rio.write_crs("EPSG:4326")

    ds = ds.rio.set_spatial_dims(
        x_dim="lon",
        y_dim="lat"
    )

    clipped = ds.rio.clip(
        shp.geometry,
        shp.crs,
        drop=True
    )

    return clipped
```

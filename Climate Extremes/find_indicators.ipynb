# Find Indicators

```
from pathlib import Path
import subprocess

PROJECT_ROOT=Path('..').resolve()
BASH_DIR=PROJECT_ROOT/'bash'
DATA_ROOT=Path('/path/to/data')
OUTPUT_ROOT=PROJECT_ROOT/'outputs'
REFERENCE_START=1980
REFERENCE_END=2000
INDICES={
'TXx':{'variable':'tasmax','script':'TXx.sh','output_folder':'TXx','needs_reference':False},
'TX90p':{'variable':'tasmax','script':'tx90p.sh','output_folder':'TX90p','needs_reference':True},
'HotDays':{'variable':'tasmax','script':'hotdays.sh','output_folder':'HotDays','needs_reference':False},
'HWFI':{'variable':'tasmax','script':'heatwavefreq.sh','output_folder':'HWFI','needs_reference':True},
'RX1day':{'variable':'pr','script':'rx1day.sh','output_folder':'RX1day','needs_reference':False},
'RX5day':{'variable':'pr','script':'rx5day.sh','output_folder':'RX5day','needs_reference':False},
'CDD':{'variable':'pr','script':'cdd.sh','output_folder':'CDD','needs_reference':False},
'CWD':{'variable':'pr','script':'cwd.sh','output_folder':'CWD','needs_reference':False},
'R95p':{'variable':'pr','script':'r95p.sh','output_folder':'R95p','needs_reference':True},
'R99p':{'variable':'pr','script':'r99p.sh','output_folder':'R99p','needs_reference':True}}
```

```
SCENARIOS=['historical','ssp126','ssp245','ssp370','ssp585']

for name,cfg in INDICES.items():
    script=BASH_DIR/cfg['script']
    for scen in SCENARIOS:
        in_dir=DATA_ROOT/scen/cfg['variable']
        out_dir=OUTPUT_ROOT/scen/cfg['output_folder']
        out_dir.mkdir(parents=True,exist_ok=True)
        for f in sorted(in_dir.glob('*.nc')):
            out=out_dir/f'{f.stem}_{name}.nc'
            cmd=['bash',str(script),str(f),str(out)]
            if cfg['needs_reference']:
                cmd.extend([str(REFERENCE_START),str(REFERENCE_END)])
            print('Running:', ' '.join(cmd))
            subprocess.run(cmd,check=True)
print('Done')
```

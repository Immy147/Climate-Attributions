# 02_Climate_Extremes_and_Trend_Analysis

This notebook is a reusable template for climate analysis workflows.

## Imports

```
from pathlib import Path
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

## Open Clean Dataset

```
ds=xr.open_dataset('processed.nc')
```

## Compute TXx

```
txx=ds['tasmax'].groupby('time.year').max()
```

## Compute RX1day

```
rx1=ds['pr'].groupby('time.year').max()
```

## Compute RX5day

```
roll=ds['pr'].rolling(time=5,min_periods=5).sum()
rx5=roll.groupby('time.year').max()
```

## Statistics

```
print(float(txx.mean()), float(txx.max()))
```

## Trend

```
ts=txx.mean(dim=['lat','lon'])
coef=np.polyfit(ts.year, ts.values,1)
print(coef)
```

## Plot

```
ts.plot(); plt.show()
```

## Export

```
txx.to_netcdf('TXx.nc')
rx1.to_netcdf('RX1day.nc')
rx5.to_netcdf('RX5day.nc')
```
